# Task Description
## Overall description

In this work, we use all the skills we have learned so far to prepare a dataset for modeling. You will conduct an Exploratory Data Analysis (EDA), build plots to understand underlying patterns, and finally prepare a clean, fully numeric dataset ready for training a machine learning model (which we will do in the next lesson).

## Detailed description

### I. Data Loading and Initial Inspection

1. Select and load a raw, uncleaned .csv dataset suitable for a classification or regression task.
2. Perform an initial inspection of the DataFrame using `.head()`, `.info()`, and `.describe()`.
3. Identify missing values in the dataset. Apply an appropriate strategy to handle them (e.g., dropping rows/columns, or imputing with mean/median/mode) and briefly explain your choice.
4. Check for and remove any duplicate rows.

### II. Exploratory Data Analysis (EDA)

1. `Univariate Analysis`: Create at least two visualizations to examine the distribution of individual key features (e.g., a Matplotlib/Seaborn histogram for numerical data, or a bar chart for categorical data).
2. `Bivariate Analysis`: Explore the relationship between your features and your assumed "target" variable. Create at least two plots (e.g., `scatter plots`, `boxplots`) to show how different features interact.
3. `Correlation`: Generate a correlation matrix for your numerical features and visualize it using a Seaborn heatmap. Identify any features that are highly correlated with each other (multicollinearity).

### III. Feature Engineering and Encoding

1. Extract valuable information from raw text or datetime columns if your dataset has them (e.g., extracting the year from a date, or using Regex to clean a string).
2. Apply appropriate encoding to all categorical variables. Use Label/Ordinal Encoding for ordered categories and One-Hot Encoding (`pd.get_dummies` or `OneHotEncoder`) for nominal categories.
3. Drop any redundant columns, ID columns, or original text columns that are no longer needed after extraction/encoding.

### IV. Final Data Preparation and Scaling

1. Identify numerical features that have vastly different scales or ranges (e.g., one column is in thousands, another is in decimals).
2. Apply a scaling technique (such as `StandardScaler` or `MinMaxScaler` from `sklearn.preprocessing`) to these specific numerical columns so they share a common scale.
3. Perform a final sanity check. Use `.info()` to prove that your DataFrame has zero missing values and that all columns are strictly numerical (int or float).
4. Save your final, ML-ready DataFrame to a new .csv file using `df.to_csv()`.
